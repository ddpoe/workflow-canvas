```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from plate_utils import load_plate, normalize, CONDITION_ORDER
%matplotlib inline
```

```python
DATA_DIR = r"C:\Users\sam\Desktop\plate_assay\data"
OUT_DIR = r"C:\Users\sam\Desktop\plate_assay\results"
PLATE = "plate1"   # change to plate2 and run again
```

```python
cells, layout = load_plate(DATA_DIR, PLATE)
print(layout["plate"], len(cells), "cells")
cells.head()
```

```python
for t in [20, 35, 50]:
    plt.figure()
    plt.hist(cells["area"], bins=80)
    plt.axvline(t, color="r")
    plt.title(f"MIN_AREA={t}: keeps {(cells['area'] >= t).sum()} cells")
    plt.show()
```

20 keeps debris, 50 cuts into real cells. **35 looks right.**

```python
# what does the threshold do to the DMSO median?
for t in [20, 35, 50]:
    n = normalize(cells[cells["area"] >= t], layout["background"])
    print(t, n.groupby("condition")["fold_change"].median().round(3).to_dict())
```
